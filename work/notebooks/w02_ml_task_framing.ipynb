## 1. My lane as an ML task (type)

**Ranking / Scoring**

We are treating this as a ranking problem via supervised classification. The system will output a probability score representing the risk of a page decaying or losing visibility. We will then use these scores to rank candidate pages, generating a prioritized queue for the SEO team to review.

## 2. Target or proxy

**Target Proxy:** The boolean label `is_declining_label`, derived from the `trend_direction` observable bucket.

This is a derived proxy from the observable trailing 90-day search metrics. In a production environment with full time-series data, a stronger target would be a future observed outcome (e.g., traffic drop over the *next* 30 days based on the *prior* 90 days), but this proxy is sufficient for our current baseline and starter modeling scope.

## 3. Success metric

**Precision@K (specifically Precision@50)**

Since the human review team has limited capacity, they will only examine the top K pages (e.g., the top 50 candidates). Therefore, we care deeply about how many true positives exist in the top 50 ranked results, rather than overall accuracy or ROC-AUC across all 30,000 pages.

## 5. Why ML beats a fixed rule here

A heuristic if-statement (e.g., `age > 180 AND impressions < 500`) relies on arbitrary, hardcoded thresholds. The real world is messier. Different content types age differently, and the relationship between position drops, CTR, and search volume is highly non-linear. An ML model (like a Random Forest) can capture these non-linear interactions and shift thresholds dynamically based on multiple interacting features, providing a much finer-grained ranking score than a blunt yes/no rule.
