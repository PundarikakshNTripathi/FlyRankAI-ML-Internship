## 1. My lane (or freestyle) and why

**Lane 2: Refresh / Content Opportunity Scoring**

I chose this lane because prioritizing human review capacity is a critical business need. By identifying which pages are underperforming, decaying, or losing visibility based on observable historical signals, we can optimally route the SEO and content teams to the most impactful opportunities.

## 2. The question: decision, action, cost of a wrong call

**The Question:** Which pages should the SEO and Content teams review first for a refresh, expansion, or protection?

**The Decision & Action:** The system will output a ranked list of candidate URLs with associated reason codes. Reviewers will examine the top-K pages and decide whether to rewrite, merge, prune, or monitor them.

**Cost of a Wrong Call:** 
1. **False Positives:** Wasting valuable human review time on pages that are naturally declining due to seasonality, or pages that are purely noise with low traffic.
2. **False Negatives:** Missing a decaying high-value page, leading to continued traffic loss that could have been mitigated.

## 3. Quick look at the data (2-3 real numbers)

Let's load the starter dataset (`data/raw/content_refresh_anonymized.csv`) to get a sense of scale. We'll look at the total number of pages and the distribution of our target proxy (`trend_direction`).

```python
import pandas as pd

# Load the local 30k starter dataset
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

total_rows = len(df)
declining_pages = len(df[df['trend_direction'] == 'down'])

print(f"Total pages in starter dataset: {total_rows:,}")
print(f"Pages with declining trend: {declining_pages:,} ({(declining_pages/total_rows)*100:.1f}%)")

# Let's also check the median impressions for the last 90 days
print(f"Median impressions (90d): {df['impressions_90d'].median():,}")
```
