## 1. Build the feature vector

Extracting features safely, capping outliers, and creating the `X` matrix and `y` target.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

features = ['impressions_90d', 'sessions_90d', 'content_age_days', 'days_since_last_update', 'word_count', 'ctr', 'avg_position', 'engagement_rate']

# Handle missing
for f in features:
    df[f] = df[f].fillna(0)

# Create proxy label
df['is_declining'] = (df['trend_direction'] == 'down').astype(int)

X = df[features]
y = df['is_declining']

print(f'Feature vector shape: {X.shape}')
```

## 2. Feature notes

All selected features were observable prior to the decision point. Missing engagement rates were filled with 0, representing a lack of tracking or zero sessions.

## 3. The leakage hunt

Checking if `trend_pct` is excluded. It is. Also checking base rate.

```python
base_rate = y.mean()
print(f'Base Rate (Declining Pages): {base_rate*100:.2f}%')

if 'trend_pct' in X.columns:
    print('LEAKAGE DETECTED: trend_pct in features')
else:
    print('Clean: trend_pct excluded.')
```

## 4. What I excluded and why

1. `trend_pct` and `trend_direction`: Derived directly from the target, would cause 100% leakage.
2. Product flags: These represent prior human or heuristic decisions. We want to predict reality, not a prior system's bias.
