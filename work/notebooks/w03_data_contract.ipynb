## 1. Unit of analysis + time window

One row = **one specific content item for a specific client**. The grain is defined by the combination of `client_id` and `content_id`.

The time window covers the trailing 90 days of performance (e.g., `impressions_90d`), representing a snapshot of historical engagement up to the date the dataset was generated.

```python
import pandas as pd

df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

# Verify grain
total_rows = len(df)
unique_grain = len(df[['client_id', 'content_id']].drop_duplicates())
print(f'Total Rows: {total_rows}, Unique Grain Rows: {unique_grain}')
assert total_rows == unique_grain, 'Grain violation detected!'
```

## 2. Fields: feature / label / context / excluded

**Features:** `impressions_90d`, `sessions_90d`, `content_age_days`, `days_since_last_update`, `word_count`, `ctr`, `avg_position`, `engagement_rate`.

**Proxy Label:** `is_declining_label` (a boolean derived from `trend_direction == 'down'`).

**Context:** `client_id`, `content_id`.

**Excluded:** `trend_direction` and `trend_pct` (these inherently contain the label information, leading to massive leakage), and any raw text fields.

## 3. Verify it with queries

Checking missing values by category to confirm no silent pattern failures.

```python
# Check missing values
missing_pct = df.isnull().mean() * 100
print('Missing values (>0%):')
print(missing_pct[missing_pct > 0].sort_values(ascending=False))
```

## 4. Data limits

This data provides a trailing 90-day snapshot. It does **not** provide full daily time-series histories for causal inference. We cannot observe the exact week a decline started, nor can we prove that a refresh will guarantee recovery. We can only rank risk based on historical snapshots.
